## Theory and assumptions

A quality detector can identify some failure evidence, such as high residual or cycle mismatch. It cannot infer missing pathology or solve modulus identifiability. The research risk register links each hazard to a verification procedure and a residual limitation. Hand-assigned costs and CPDs are educational assumptions. They must not become calibrated medical risk probabilities through a change of label.

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.data.synthetic import make_pair
from elastograde.evaluation.robustness import corrupt
from elastograde.inference.pipeline import infer
s=make_pair(seed,kind='compression');rows=[]
for name,kind,severity in [('clean','noise',0),('noisy','noise',1),('dropout','dropout',.3)]:
 r=infer(corrupt(s,kind,severity));rows.append((name,r['quality'],r['decision']))
print(json.dumps(rows,indent=2))
plt.bar([r[0] for r in rows],[r[1]['residual_mean'] for r in rows]);plt.title('synthetic_fixture: observed quality failures');plt.ylabel('Mean RF residual');plt.show()
```

## Interpretation and failure cases

Threshold actions are demonstrative. A missed failure is useful research evidence and should stay in the failure atlas instead of being hidden by removing a difficult sample.
