## Theory and assumptions

For small deformation, $\epsilon_{zz}=\partial u_z/\partial z$, $\epsilon_{xx}=\partial u_x/\partial x$ and $\epsilon_{xz}=\frac12(\partial u_x/\partial z+\partial u_z/\partial x)$. Convert displacement and coordinates consistently. Engineering shear is twice tensor shear. A least-squares derivative reduces noise at the cost of spatial resolution. Freehand strain without a known stress field cannot determine an absolute Young modulus.

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.data.synthetic import make_pair
from elastograde.mechanics.strain import strain
from elastograde.mechanics.features import mechanical_features
s=make_pair(seed,kind='compression');v=strain(s.displacement,s.metadata['spacing_mm'],'least_squares')
print('Expected compression strain -0.025; observed median:',np.median(v['axial']))
print(mechanical_features(v['axial'],s.valid,s.roi,s.metadata['spacing_mm']))
plt.imshow(v['axial'],cmap='coolwarm',vmin=-.03,vmax=0);plt.colorbar();plt.title('synthetic_fixture: analytic axial strain');plt.show()
```
