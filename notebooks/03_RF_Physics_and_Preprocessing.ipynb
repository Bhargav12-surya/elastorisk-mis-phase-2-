## Theory and assumptions

RF is a signed oscillatory signal. An analytic signal along the axial direction is $s_a=s+i\mathcal H(s)$, and its envelope is $|s_a|$. Display B-mode is $20\log_{10}(A/A_{max})$ clipped to a chosen dynamic range. These transformations are not interchangeable: envelope and display compression discard information useful for phase-sensitive motion estimation. Pulse-echo axial spacing is $c/(2f_s)$ only when the raw sampling model applies; lateral spacing is independent metadata.
