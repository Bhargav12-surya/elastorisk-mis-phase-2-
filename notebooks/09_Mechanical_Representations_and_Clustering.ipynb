## Theory and assumptions

Standardization and dimensionality reduction are fitted on training features. A Gaussian mixture supplies soft membership in descriptive mechanical clusters. BIC balances fit and model complexity; it does not discover the true number of clinical grades. Bootstrap adjusted Rand agreement measures clustering stability under resampling. A nuisance-prediction probe asks whether embeddings encode scanner or compression conditions.

```python
from pathlib import Path
import os,json
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'configs').exists():ROOT=ROOT.parent
mode='demo'
data_root=ROOT/'data/raw'
config_path=ROOT/'configs/demo.yaml'
seed=42
max_samples=8
device='cpu'
output_dir=ROOT/'reports/notebooks'/'09_Mechanical_Representations_and_Clustering'
output_dir.mkdir(parents=True,exist_ok=True)
print('Mode:',mode,'| Data origin: synthetic_fixture')
```

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.models.representations import MechanicalPhenotypes,nuisance_probe
rng=np.random.default_rng(seed);X=np.r_[rng.normal(-1,.4,(25,4)),rng.normal(1,.4,(25,4))]
m=MechanicalPhenotypes().fit(X,['train']*len(X),components=[2,3]);p=m.predict(X)
print(m.stability(X,repeats=4));print(p['names'][:4])
plt.scatter(X[:,0],X[:,1],c=p['probabilities'].argmax(1),cmap='viridis');plt.title('synthetic_fixture: descriptive clusters, not grades');plt.show()
print(nuisance_probe(X[:40],X[:40,0],X[40:],X[40:,0]))
```

## Interpretation and failure cases

The nuisance probe intentionally predicts a feature-derived nuisance, demonstrating how a representation can contain an unwanted shortcut. Cluster colors are arbitrary labels.
