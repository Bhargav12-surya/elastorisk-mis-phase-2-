## Theory and assumptions

Moralization connects co-parents and removes arrow directions. Assigning each Bayesian CPD to its family clique preserves the factor product and therefore the joint distribution. The undirected graph can lose some independencies; moralization is not triangulation or causal discovery. A decision network adds actions and utilities to uncertain states. The Bayes action minimizes expected loss: $a^*=\arg\min_a\sum_sL(a,s)P(s\mid e)$.

For reliability probability 0.8, the illustrative acceptance loss is $10(1-0.8)=2$, reacquisition loss is 1, and review loss is $2(0.2)+0.5(0.8)=0.8$. Review wins under these assumed costs.

```python
from elastograde.pgm.bayesian_quality import illustrative_network
from elastograde.pgm.moralization import verify
from elastograde.pgm.decisions import choose,sensitivity
print(verify(illustrative_network()));print(choose(.8))
ps=np.linspace(0,1,40);rows=sensitivity(ps,[5,10,20]);names={'accept_mechanics':0,'reacquire':1,'expert_review':2}
for cost in [5,10,20]:plt.step(ps,[names[r['action']] for r in rows if r['false_accept_cost']==cost],label=f'False-accept cost {cost}')
plt.yticks([0,1,2],list(names));plt.xlabel('Assumed reliability probability');plt.legend();plt.title('Educational decision sensitivity');plt.show()
```
