## Theory and assumptions

The HMM assumes a hidden Markov state and conditionally independent observations. Filtering uses observations up to time $t$; smoothing uses the whole sequence; Viterbi selects the most probable complete state path. They answer different questions. Sequence boundaries must reset the initial state. A collection of unrelated compression pairs cannot be converted into a temporal experiment by sorting filenames.
