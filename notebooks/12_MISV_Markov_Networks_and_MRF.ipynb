## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.pgm.markov_network import separated
from elastograde.pgm.mrf import icm,exact_map,energy
print('a independent of c given b:',separated([('a','b'),('b','c')],{'a'},{'c'},{'b'}))
rng=np.random.default_rng(seed);truth=np.zeros((18,18),int);truth[7:11,7:11]=1
obs=truth+rng.normal(0,.35,truth.shape);unary=np.stack([obs**2,(obs-1)**2],-1)
labels,history=icm(unary,beta=.3)
fig,axes=plt.subplots(1,3,figsize=(10,3))
for ax,img,title in zip(axes,[truth,unary.argmin(-1),labels],['Toy reference','Independent unary','Conditional Potts']):ax.imshow(img,cmap='viridis');ax.set_title(title)
plt.suptitle('synthetic_fixture');plt.show();print('Energy:',history)
small=unary[:2,:2];print('Tiny exact MAP:',exact_map(small)[1])
```

## Interpretation and failure cases

Improved visual smoothness is not sufficient evidence: evaluate small-region sensitivity and boundary error. Large beta can favor a uniform image even when a small true inclusion exists.
