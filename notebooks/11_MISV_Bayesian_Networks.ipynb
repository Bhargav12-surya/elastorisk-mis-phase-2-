## Theory and assumptions

A Bayesian network factorizes a joint distribution as $P(V)=\prod_iP(V_i\mid Pa_i)$. In the illustrative quality graph, domain influences quality and reliability, while quality influences residual, cycle evidence and uncertainty. Inference sums over unobserved variables. Variable elimination reuses factors instead of enumerating every assignment. The graph is an assumed probabilistic structure, not a demonstrated causal model.

Representation learning can introduce a latent quality state explaining several observed binary features. EM alternates posterior membership and parameter updates; latent state names can permute, so state 1 is not intrinsically good quality.

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.pgm.bayesian_quality import illustrative_network
from elastograde.pgm.latent_quality import BernoulliMixture
bn=illustrative_network();e={'residual':0,'cycle':0}
print('Variable elimination:',bn.query('reliable',e));print('Enumeration:',bn.enumerate_query('reliable',e))
rng=np.random.default_rng(seed);latent=rng.integers(0,2,80);X=(rng.random((80,3))<np.where(latent[:,None]==1,.85,.2)).astype(int)
em=BernoulliMixture().fit(X,['train']*80)
plt.plot(em.history);plt.title('synthetic_fixture: latent mixture fit');plt.ylabel('Mean log likelihood');plt.show()
print('Learned emissions:',em.p)
```

## Interpretation and failure cases

Network CPDs are assumed educational numbers. EM emissions are learned from synthetic observations. Neither output is a clinically calibrated reliability estimate.
