# Uncertainty and Selective Prediction

**Question:** Does rejection reduce error while preserving useful coverage?

**Prerequisites:** install the project with notebook extras; run from the repository root. This notebook is self-contained in bounded demo mode.

## Theory and assumptions

Residual and cycle mismatch are uncalibrated quality proxies. Epistemic disagreement needs multiple independently fitted models. Calibration needs held-out groups. Selective risk is the error among accepted cases; coverage is the accepted fraction. Reporting error alone rewards rejecting nearly everything. Conformal group-level intervals require a clear coverage event and exchangeability; their coverage is not guaranteed under arbitrary scanner shift.

```python
from pathlib import Path
import os,json
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'configs').exists():ROOT=ROOT.parent
mode='demo'
data_root=ROOT/'data/raw'
config_path=ROOT/'configs/demo.yaml'
seed=42
max_samples=8
device='cpu'
output_dir=ROOT/'reports/notebooks'/'10_Uncertainty_and_Selective_Prediction'
output_dir.mkdir(parents=True,exist_ok=True)
print('Mode:',mode,'| Data origin: synthetic_fixture')
```

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.uncertainty.selective import risk_coverage
from elastograde.uncertainty.calibration import conformal_radius,TemperatureScaler,reliability
rng=np.random.default_rng(seed);errors=rng.uniform(0,1,30);uncertainty=errors+rng.normal(0,.15,30)
r=risk_coverage(errors,uncertainty)
plt.plot(r['coverage'],r['risk']);plt.xlabel('Retained independent groups');plt.ylabel('Mean toy error');plt.title('synthetic_fixture: selective risk');plt.show()
print('Group nonconformity radius:',conformal_radius(errors,.1))
logits=rng.normal(size=(40,3));labels=rng.integers(0,3,40)
t=TemperatureScaler().fit(logits,labels,['calibration']*40)
print('Temperature:',t.temperature);print(reliability(t.predict(logits),labels))
```

## Interpretation and failure cases

The uncertainty-error association is deliberately generated for teaching, not measured model calibration. Evaluating the fitted temperature on the same calibration sample is a fit diagnostic, not held-out validation.
