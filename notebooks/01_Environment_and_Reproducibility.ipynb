## Theory and assumptions

A seed determines pseudo-random numbers, but reproducibility also depends on numerical libraries, hardware and deterministic kernels. A run records its dependency versions, source hash, configuration and source checksums. GPU interpolation gradients can be nondeterministic. Reproducibility should therefore specify a numerical tolerance rather than promise identical bytes across all platforms.
