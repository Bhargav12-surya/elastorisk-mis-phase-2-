**Question:** Does local correlation recover known subpixel motion?

**Prerequisites:** install the project with notebook extras; run from the repository root. This notebook is self-contained in bounded demo mode.

## Theory and assumptions

Windowed normalized cross-correlation subtracts local means and divides covariance by local standard deviations. An integer displacement search locates the strongest correlation peak. A parabolic fit around that peak provides a local subpixel estimate, but its assumptions can fail with repeated RF phase peaks or decorrelation. The baseline searches signed correlation and declares invalid borders rather than interpreting padded values as tissue.

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.data.synthetic import make_pair
from elastograde.baselines.ncc import estimate
from elastograde.baselines.regularized import refine
from elastograde.evaluation.metrics import displacement_metrics
s=make_pair(seed);u,c,m=estimate(s.pre,s.post);v=refine(u,c)
print('NCC:',displacement_metrics(u,s.displacement,m))
print('Regularized:',displacement_metrics(v,s.displacement,m))
fig,axes=plt.subplots(1,2,figsize=(9,3))
axes[0].imshow(u[1],cmap='coolwarm');axes[0].set_title('Axial displacement pixels')
axes[1].imshow(c,cmap='viridis');axes[1].set_title('NCC proxy confidence')
plt.suptitle('synthetic_fixture');plt.show()
```

## Interpretation and failure cases

An accurate constant translation does not establish strain accuracy. Strain differentiates motion, amplifying small spatial errors. The regularizer is an original simple quadratic smoother, not a reproduction of a named published solver.
