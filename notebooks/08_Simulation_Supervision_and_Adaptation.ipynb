## Theory and assumptions

Pure self-supervision uses observed pairs. Simulation supervision uses reference displacement generated by a simulator. Hybrid learning starts from a simulation-supervised checkpoint and adapts using unlabeled acquired RF. Calling all three label-free would hide a substantial difference in information access. Evaluate them on identical held-out groups with comparable tuning budgets.

```python
from pathlib import Path
import os,json
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'configs').exists():ROOT=ROOT.parent
mode='demo'
data_root=ROOT/'data/raw'
config_path=ROOT/'configs/demo.yaml'
seed=42
max_samples=8
device='cpu'
output_dir=ROOT/'reports/notebooks'/'08_Simulation_Supervision_and_Adaptation'
output_dir.mkdir(parents=True,exist_ok=True)
print('Mode:',mode,'| Data origin: synthetic_fixture')
```

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.config import load
from elastograde.training.engine import train
base=load(ROOT/'configs/demo.yaml');base.update(samples=8,epochs=1)
supervised=dict(base,output=str(output_dir/'supervised'),regime='simulation_supervised',loss=dict(base['loss'],rf=0.,env=0.,sup=1.))
a=train(supervised)
hybrid=dict(base,output=str(output_dir/'hybrid'),regime='hybrid',pretrained=str(output_dir/'supervised/best.pt'))
b=train(hybrid)
print({'supervised':a['status'],'adaptation':b['status'],'data_origin':'synthetic_fixture','real_domain_adaptation':'not_evaluated'})
```
