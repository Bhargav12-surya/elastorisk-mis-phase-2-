# Self Supervised Learning

**Question:** Can a compact flow network train without motion labels?

**Prerequisites:** install the project with notebook extras; run from the repository root. This notebook is self-contained in bounded demo mode.

## Theory and assumptions

A backward sampling field reconstructs reference RF as $I_m(x+u(x))$. The self-supervised objective combines RF/envelope similarity, forward-backward consistency, anisotropic smoothness and a Jacobian penalty. None uses reference motion when the supervised weight is zero. A low reconstruction loss can still accompany incorrect motion; validate on independent known-motion data. The candidate uses shared encoders, local feature correlation and coarse-to-fine residual refinement.

```python
from pathlib import Path
import os,json
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if not (ROOT/'configs').exists():ROOT=ROOT.parent
mode='demo'
data_root=ROOT/'data/raw'
config_path=ROOT/'configs/demo.yaml'
seed=42
max_samples=8
device='cpu'
output_dir=ROOT/'reports/notebooks'/'07_Self_Supervised_Learning'
output_dir.mkdir(parents=True,exist_ok=True)
print('Mode:',mode,'| Data origin: synthetic_fixture')
```

## Experiment

Inputs: seeded synthetic fixtures or explicitly identified educational probability tables. Outputs: printed numerical evidence, saved run artifacts where applicable, and plots. Real-data-only claims remain unavailable until source archives and metadata are verified.

```python
from elastograde.config import load
from elastograde.training.engine import train
cfg=load(ROOT/'configs/demo.yaml');cfg.update(output=str(output_dir/'training'),epochs=1,samples=8)
summary=train(cfg)
print(summary)
rows=[json.loads(x) for x in (output_dir/'training/history.jsonl').read_text().splitlines()]
plt.plot([r['epoch'] for r in rows],[r['train_loss'] for r in rows],'o-',label='train');plt.plot([r['epoch'] for r in rows],[r['validation_loss'] for r in rows],'s-',label='validation');plt.legend();plt.title('synthetic_fixture: one-epoch smoke training');plt.show()
```
